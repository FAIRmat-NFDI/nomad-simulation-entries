```python
import requests
```

```python
from copy import deepcopy
```

```python
from time import sleep
```

```python
URL="https://nomad-lab.eu/backend/api/v1/entries/query"
```

```python
BASE_QUERY = {
  "owner": "visible",
  "query": {
    "quantities:any": [
      "run.program.version"
    ]
  },
  "pagination": {
    "order_by": "upload_create_time",
    "order": "desc",
    "page_size": 1
  },
  "required": {
    "include": [
      "entry_id",
    ]
  }
}
```

```python
result = requests.post(URL, json=BASE_QUERY)
```

```python
result.raise_for_status()
```

```python
result.json()['pagination']['total']
```

```python
def get_total_usage_per_quantity(quantity_path, base_url = URL, query=BASE_QUERY):
    query = deepcopy(query)
    query["query"]["quantities:any"] = [quantity_path]
    result = requests.post(URL, json=query)
    result.raise_for_status()
    return result.json()['pagination']['total']
```

```python
# this is what we use in the Simulations App
all_simulations = get_total_usage_per_quantity("results.method.simulation.program_name")
```

```python
quantites_to_test = ['run.calculation.dos_electronic.energy_fermi', "run.method.atom_parameters", "run.method.x_vasp_incar_in"]
```

```python
print(f"|{'quantity path':30s}|{'total number':>20s}|{'percentage':>20s}|")
print(74*"-")
for quantity in quantites_to_test:
    n_tot = get_total_usage_per_quantity(quantity)
    print(f"|{quantity[-30:]:30s}|{n_tot:20d}|{n_tot/all_simulations*100:20.3f}|")
    sleep(0.5)
```
